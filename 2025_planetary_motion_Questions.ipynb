# Planetary motion
 In this project you are going to simulate how planets move around the sun. Everything starts with the Newtonian
equations of motion for the positions of each object $i$:
\begin{equation}
m_i\mathbf{ \ddot r_i}=G\sum_{j}m_im_j \frac{\mathbf{ r_j} -\mathbf{ r_i}}{|\mathbf{ r_j} -\mathbf {r_i}|^3},
\end{equation}
where $m_j$ are the masses of the other objects, $\mathbf{r_j}$ are their positions and $G=6.6743\times 10^{-11}\; m^3/kg s^2$ is the universal gravitational constant. Notice that this is an ODE of second order, as it involves the second derivative of the position vector.
Gravitation is a conservative force, i.e., the total energy of the system $E=T+V$ does not vary in time, where $E$ is given by:
\begin{equation}
E=\frac{1}{2}\sum_i m_i |\mathbf{ \dot r_i}|^2-G\sum_{i,j}\frac{m_im_j}{|\mathbf{ r_j} -\mathbf {r_i}|}
\end{equation}

It is theoretically difficult to find a numerical intergration scheme that is consistent with energy conservation. Systems that conserve their energy are called Hamiltonian systems in classical mechanics. 
An algorithm that can integrate a Hamiltonian dynamical system is called a _symplectic integrator_. The standard algorithms that we have discussed during the lectures, such as the Euler method and the Runge-Kutta method, are not symplectic integrators. Instead, we are going to use the Newton-Stormer-Verlet algorithm, which superficially resembles Euler. Suppose that our equations of motion can be re-written as standard Newton equations, such that we have a system of 2 coupled ODES of first order:
\begin{align}
\mathbf{\dot r}&=\mathbf{v}\\
\mathbf{\dot v}&=\frac{\mathbf{F}}{m}
\end{align}.
Then, in the Newton-Stormer-Verlet algorithm, we first update the velocity, and then, with the updated velocity, we update the positions:
\begin{align}
\mathbf{v_{k+1}}&=\mathbf{v_k}+\frac{\mathbf{F_k}}{m}dt\\
\mathbf{r_{k+1}}&=\mathbf{r_k}+\mathbf{v_{k+1}}dt
\end{align}
